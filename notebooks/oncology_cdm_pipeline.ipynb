```python
dm = df.pivot_table(
    index="patientId",
    columns="clinicalAttributeId",
    values="value",
    aggfunc="first"
).reset_index()

dm.columns = dm.columns.astype(str).str.upper()
```

```python
dm = dm.rename(columns={
    "PATIENTID": "USUBJID"
})

dm = dm[[
    "USUBJID",
    "CANCER_TYPE",
    "CANCER_TYPE_DETAILED",
    "FRACTION_GENOME_ALTERED",
    "ANEUPLOIDY_SCORE",
    "MSI_SCORE_MANTIS"
]]
```

```python
import pandas as pd

num_cols = ["FRACTION_GENOME_ALTERED", "ANEUPLOIDY_SCORE", "MSI_SCORE_MANTIS"]

for col in num_cols:
    dm[col] = pd.to_numeric(dm[col], errors="coerce")
```

```python
import numpy as np

np.random.seed(42)

ae = pd.DataFrame({
    "USUBJID": dm["USUBJID"],
    "AE_TERM": np.random.choice(
        ["head pain", "loose stool", "fatigue", "fever", "nausea"],
        len(dm)
    )
})
```

```python
meddra_map = {
    "head pain": "Headache",
    "loose stool": "Diarrhea",
    "fever": "Pyrexia",
    "nausea": "Nausea",
    "fatigue": "Fatigue"
}

ae["AE_PT"] = ae["AE_TERM"].map(meddra_map)
```

```python
tumor = pd.DataFrame({
    "USUBJID": dm["USUBJID"],
    "BASELINE": np.random.uniform(20, 100, len(dm))
})

tumor["CURRENT"] = tumor["BASELINE"] * np.random.uniform(0.4, 1.6, len(dm))
```

```python
def recist(row):
    change = ((row["CURRENT"] - row["BASELINE"]) / row["BASELINE"]) * 100

    if row["CURRENT"] == 0:
        return "CR"
    elif change <= -30:
        return "PR"
    elif change >= 20:
        return "PD"
    else:
        return "SD"

tumor["RESPONSE"] = tumor.apply(recist, axis=1)
```

```python
orr = tumor["RESPONSE"].isin(["CR", "PR"]).mean()
print("ORR:", round(orr, 3))
```

```python
from lifelines import KaplanMeierFitter

surv = pd.DataFrame({
    "USUBJID": dm["USUBJID"],
    "TIME": np.random.exponential(24, len(dm)),  # months
    "EVENT": np.random.binomial(1, 0.4, len(dm)) # death flag
})
```

```python
kmf = KaplanMeierFitter()

kmf.fit(
    durations=surv["TIME"],
    event_observed=surv["EVENT"]
)

kmf.plot()
```

```python
final = dm.merge(ae, on="USUBJID").merge(tumor, on="USUBJID").merge(surv, on="USUBJID")
final.head()
```
